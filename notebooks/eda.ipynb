Then we have the different possitions, first we distinghuish between:
- Offense
    - Quarterback (QB): leader of the offense, calls plays he is the floor general
    - Running Back (RB): Must be fast, explosive and great lower body strenght, usually not very tall
    - Fullback (FB): Located behind the middle of the line, in addition to running, he blocks open holes on running plays and sometimes catches short passes
    - Offensive line: their job is to block ad allow the QB time to run a play, has 5 offensive linemen:
        - Left Tackle (LT)
        - Left Guard (LG)
        - Right Guard (RG)
        - Right Tackle (RT)
    - Wide receivers (WR): there are two of them, they receive the passes. Fast and light. Usualyy the ones scoring TouchDowns
    - Tight End (TE): Combination of receiver and offensive lineman, they can block or separete from the line to catch passes

- Defense:
    - Defensive line: made of 3 or 4 players who line up in front of the offensive line, they are the first line of defense. Usually they are two Defensive Ends (DE) on either side and one or two Defensice Tackles (DT). Their objecive is to get rid of the offensive linemens block and tackle the ball carrier
    - LineBacker (LB): responsible for backing up the defemsve line. Depending on defensive alignment there may be 3 or 4 players of this position on the field
    - Cornerback (CB): usually two players (fastest of the defensive line) they cover wide receivers, but also blitz the quarterback
    - Safety: divided into two postions Strong Safety(SS) and Free Safety (FS) they must be fast strong and good tacklers

- Special Teams:
    - Kicker
    - Holder
    - Long snapper
    - Punter
    - Return specialist
